```python
def print_ner_from_json(file_path):
    try:
        with open(file_path, "r", encoding="utf-8") as f:
            data = json.load(f)

            print(f"\n{'='*50}\nNER from {os.path.basename(file_path)}\n{'='*50}")

            for article_id, article_data in data.items():
                print(f"\nid: {article_id}")

                if "entities" in article_data:
                    for entity in article_data["entities"]:
                        text = entity.get("text_span", "N/A")
                        label = entity.get("label", "N/A")
                        print(f"  - **{label}**: {text}")
                else:
                    print("No entities found")

    except FileNotFoundError:
        print(f"Error: File not found - {file_path}")
    except json.JSONDecodeError:
        print(f"Error: Invalid JSON format in {file_path}")
    except PermissionError:
        print(f"Error: Permission denied for {file_path}")

# Process each JSON file
for json_file in json_files:
    print_ner_from_json(json_file)
```

```python
import json

def normalize_title(title):
    return title.strip().lower()

with open(r"C:\Users\samue\OneDrive\Desktop\ataupd2425-gainer\Annotations\Train\platinum_quality\json_format\train_platinum.json", "r", encoding="utf-8") as f:
    platinum = json.load(f)

with open(r"C:\Users\samue\OneDrive\Desktop\ataupd2425-gainer\Annotations\Train\gold_quality\json_format\train_gold.json", "r", encoding="utf-8") as f:
    gold = json.load(f)

platinum_titles = set()
for article in platinum.values():
    title = article["metadata"]["title"]
    platinum_titles.add(normalize_title(title))

merged = platinum.copy()

for article_id, article_data in gold.items():
    title = article_data["metadata"]["title"]
    norm_title = normalize_title(title)
    if norm_title not in platinum_titles:
        merged[article_id] = article_data

with open("merged_dataset.json", "w", encoding="utf-8") as f:
    json.dump(merged, f, ensure_ascii=False, indent=2)

print("Mergeddataset saved as merged_dataset.json")
```

```python
#debug of the merged dataset
def print_json_statistics(json_file):
    with open(json_file, "r", encoding="utf-8") as f:
        data = json.load(f)

    num_documents = len(data)
    total_entities = 0
    total_relations = 0
    for article in data.values():
        total_entities += len(article.get("entities", []))
        total_relations += len(article.get("relations", []))

    avg_entities = total_entities / num_documents if num_documents > 0 else 0
    avg_relations = total_relations / num_documents if num_documents > 0 else 0

    print("Statistics for", json_file)
    print("Num of Documents:\t", num_documents)
    print("Total Entities:\t\t", total_entities)
    print("Avg Entities per Doc:\t", avg_entities)
    print("Total Relations:\t", total_relations)
    print("Avg Rels per Doc:\t", avg_relations)

if __name__ == "__main__":
    print_json_statistics(r"C:\Users\samue\OneDrive\Desktop\ataupd2425-gainer\merged_dataset.json")
```

```python
#Count the nmber of tokens in the abstracts of the articles in the gold/platinum/silver/bronze collection
import json
from transformers import AutoTokenizer

def main():
    #load the tokenizer
    tokenizer = AutoTokenizer.from_pretrained("microsoft/deberta-v3-large")
    
    #load the ds
    with open("/kaggle/input/gold-ds/train_gold.json", "r") as f:
        articles = json.load(f)
    
    max_tokens = 0
    total_tokens = 0
    num_docs = 0
    count_over_600 = 0
    count_over_700 = 0
    count_over_800 = 0

    for doc_id, info in articles.items():
        abstract = info["metadata"]["abstract"]
        # Tokenize
        encoding = tokenizer(abstract, add_special_tokens=True)
        num_tokens = len(encoding["input_ids"])
        
        total_tokens += num_tokens
        num_docs += 1
        if num_tokens > max_tokens:
            max_tokens = num_tokens
        
        if num_tokens > 600:
            count_over_600 += 1
        if num_tokens > 700:
            count_over_700 += 1
        if num_tokens > 800:
            count_over_800 += 1

        print(f"Document ID: {doc_id} - Numero di token: {num_tokens}")
    
    avg_tokens = total_tokens / num_docs if num_docs > 0 else 0
    print(f"\nMaxtoken: {max_tokens}")
    print(f"Mean token for document: {avg_tokens:.2f}")
    print(f"Number of docs with token > 600: {count_over_600}")
    print(f"Number of docs with token > 700: {count_over_700}")
    print(f"Number of docs with token > 800: {count_over_800}")

if __name__ == "__main__":
    main()
```
